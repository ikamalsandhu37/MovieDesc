```python
import torch
import torchvision.models.video as models
import torchvision.transforms as transforms
from torchvision.io import read_video
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import os

# Using ResNet-based 3D CNN (I3D equivalent)
model = models.r3d_18(pretrained=True)  # You can also use models.mc3_18

# Modify the final layer for 2 classes (e.g., "orange_slice_action" vs. "background")
num_classes = 2
model.fc = nn.Linear(in_features=model.fc.in_features, out_features=num_classes)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

#  Prepare Video Data
class VideoDataset(Dataset):
    def __init__(self, video_dir, transform=None):
        self.video_dir = video_dir
        self.video_files = [f for f in os.listdir(video_dir) if f.endswith(".mp4")]
        self.transform = transform

    def __len__(self):
        return len(self.video_files)

    def __getitem__(self, idx):
        video_path = os.path.join(self.video_dir, self.video_files[idx])
        video, _, _ = read_video(video_path, pts_unit="sec")  # Read as (T, H, W, C)
        
        # Convert to float and permute to (C, T, H, W) format
        video = video.float().permute(3, 0, 1, 2) / 255.0
        
        if self.transform:
            video = self.transform(video)
        
        label = 1  # Assuming all videos belong to "orange_slice_action" (change as needed)
        return video, torch.tensor(label, dtype=torch.long)

# Define transformations
video_transforms = transforms.Compose([
    transforms.Resize((112, 112)),  # Resize to model input size
    transforms.Normalize(mean=[0.5], std=[0.5])  # Normalize pixel values
])

video_path = "C:Users/jagji/OneDrive/Desktop/video_downloads/video_2.mp4"
dataset = VideoDataset(video_path, transform=video_transforms)
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)

# Train the Model
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

num_epochs = 5

for epoch in range(num_epochs):
    model.train()
    total_loss = 0.0

    for videos, labels in dataloader:
        videos, labels = videos.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(videos)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()
    
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {total_loss / len(dataloader):.4f}")


# Save the Fine-Tuned Model
torch.save(model.state_dict(), "fine_tuned_I3D.pth")
print("Model saved successfully!")
```
